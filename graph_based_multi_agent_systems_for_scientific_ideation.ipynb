```python
def planner_agent(user_query):

    prompt = f"""
    Analyze this research topic:

    {user_query}

    Give:
    1. Main research domain
    2. Important subtopics
    3. Required technologies
    4. Research workflow
    """

    result = generator(
        prompt,
        max_length=300
    )

    return result[0]['generated_text']
```

```python
query = """
Task-feature-based training for tool use and graph-based multi-agent frameworks for scientific ideation
"""

response = planner_agent(query)

print(response)
```

```python
def search_papers(query, max_results=5):

    search = arxiv.Search(
        query=query,
        max_results=max_results,
        sort_by=arxiv.SortCriterion.Relevance
    )

    papers = []

    for result in search.results():

        paper = {
            "title": result.title,
            "authors": [a.name for a in result.authors],
            "summary": result.summary,
            "published": result.published,
            "pdf_url": result.pdf_url
        }

        papers.append(paper)

    return papers
```

```python
import arxiv

def search_papers(query, max_results=5):
    # Initialize the arxiv client with more retries and a longer delay to handle persistent rate limiting
    client = arxiv.Client(num_retries=20, delay_seconds=10) # Increased num_retries and delay_seconds again

    search = arxiv.Search(
        query=query,
        max_results=max_results,
        sort_by=arxiv.SortCriterion.Relevance
    )

    papers = []

    for result in client.results(search):

        paper = {
            "title": result.title,
            "authors": [a.name for a in result.authors],
            "summary": result.summary,
            "published": result.published,
            "pdf_url": result.pdf_url
        }

        papers.append(paper)

    return papers

papers = search_papers(
    "multi-agent systems for scientific discovery"
)

print("Total Papers:", len(papers))
```

```python
for i, paper in enumerate(papers):

    print(f"\n📄 PAPER {i+1}")
    print("TITLE:", paper['title'])
    print("AUTHORS:", paper['authors'])
    print("PDF:", paper['pdf_url'])
```

PHASE 3 — Scientific Paper Summarizer Agent

```python
def summarize_paper(abstract):

    prompt = f"""
    Read this scientific abstract carefully.

    ABSTRACT:
    {abstract}

    Give:
    1. Main idea
    2. Key contribution
    3. Important technologies
    4. Limitations

    Keep the explanation simple and clear.
    """

    result = generator(
        prompt,
        max_length=300
    )

    return result[0]['generated_text']
```

```python
abstract = papers[0]['summary']

summary = summarize_paper(abstract)

print(summary)
```

PHASE 4 — Research Gap Detector Agent

```python
def detect_research_gap(paper_summaries):

    combined_text = "\n\n".join(paper_summaries)

    prompt = f"""
    Analyze these research paper summaries carefully.

    SUMMARIES:
    {combined_text}

    Identify:

    1. Common limitations
    2. Missing research areas
    3. Unsolved technical problems
    4. Future research opportunities

    Give detailed research gaps.
    """

    result = generator(
        prompt,
        max_length=400
    )

    return result[0]['generated_text']
```

```python
paper_summaries = []

for paper in papers:

    summary = summarize_paper(paper['summary'])

    paper_summaries.append(summary)

print("Total Summaries:", len(paper_summaries))
```

```python
gaps = detect_research_gap(paper_summaries)

print(gaps)
```

NEXT PHASE = IDEA GENERATOR

```python
def generate_research_idea(gaps):

    prompt = f"""
    You are an advanced AI research scientist.

    Based on these research gaps:

    {gaps}

    Generate:

    1. A novel research idea
    2. Proposed AI architecture
    3. Suggested technologies
    4. Possible experiments
    5. Potential paper title

    Make the idea innovative and realistic.
    """

    result = generator(
        prompt,
        max_length=500
    )

    return result[0]['generated_text']
```

```python
idea = generate_research_idea(gaps)

print(idea)
```
